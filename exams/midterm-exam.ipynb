## Problem 1. The Pitot-static tube: how an aircraft measures its own speed

An aircraft has no speedometer. It measures its speed by measuring a pressure difference: a **Pitot-static tube** faces into the flow, so that its central port reads the **stagnation pressure** $p_0$ at a point where the flow has been brought to rest, while side ports read the **static pressure** $p$ of the undisturbed stream. In this problem you will build the instrument's calibration and find out where it starts to lie.

### The incompressible reading:

Applying the Bernoulli equation along the streamline that ends at the stagnation point, with no change in elevation:

$$p_0 = p + \tfrac{1}{2}\rho V^{2} \qquad \Longrightarrow \qquad V = \sqrt{\frac{2\,\Delta p}{\rho}}, \qquad \Delta p = p_0 - p$$

The instrument measures $\Delta p$. To turn it into a speed it needs a density, and the density of the air depends on the altitude.

### The atmosphere:

Use the standard atmosphere of the troposphere, as in the lectures:

$$T(z) = T_a - \beta z, \qquad p(z) = p_a\left(1 - \frac{\beta z}{T_a}\right)^{g/(R\beta)}, \qquad \rho = \frac{p}{R\,T}$$

with $T_a = 288.15\,\rm K$, $p_a = 101.33\,\rm kPa$, $\beta = 0.0065\,\rm K/m$, $R = 286.9\,\rm J/(kg\,K)$ and $g = 9.81\,\rm m/s^2$, valid up to $z = 11\,\rm km$.

### The compressible reading:

Air is a gas, and bringing it to rest at the tip of the probe compresses it. For an isentropic deceleration of a perfect gas with $\gamma = 1.4$, the exact stagnation relation is

$$\frac{p_0}{p} = \left(1 + \frac{\gamma-1}{2}\,M^{2}\right)^{\gamma/(\gamma-1)}, \qquad M = \frac{V}{a}, \qquad a = \sqrt{\gamma\,R\,T}$$

where $M$ is the Mach number and $a$ the speed of sound. The Bernoulli result above is the limit of this expression for $M \ll 1$.

## Problem 2. Streamlines, pathlines and streaklines in a precessing jet

Some accreting compact objects launch a narrow jet of gas whose axis does not stay fixed, but **precesses** slowly around a cone. Radio maps of such sources show a striking **corkscrew** pattern, which this problem addresses. The model below is idealised, but the numbers are of the right order for a Galactic jet source.

### The model:

The jet material is **ballistic**: each blob is ejected from the centre at speed $v$ and then travels in a straight line, with no further acceleration. What changes with time is only the **direction of ejection**, which sweeps around a cone of half-angle $\theta$ about the precession axis $x$:

$$\hat{\mathbf{n}}(t) = \left(\cos\theta,\;\; \sin\theta\,\cos\omega t,\;\; \sin\theta\,\sin\omega t\right), \qquad \omega = \frac{2\pi}{P}$$

A blob ejected at time $t_0$ therefore has position

$$\mathbf{x}(t;t_0) = v\,(t - t_0)\;\hat{\mathbf{n}}(t_0)$$

Because material found at a distance $r$ from the centre was ejected a time $r/v$ earlier, the **velocity field** of the jet is

$$\mathbf{v}(\mathbf{x},t) = v\,\hat{\mathbf{n}}\!\left(t - \frac{r}{v}\right), \qquad r = |\mathbf{x}|$$

Use the following parameters:

| Quantity | Symbol | Value |
|---|---|---|
| Jet speed | $v$ | $0.25\,c$ |
| Precession period | $P$ | $160\,\rm days$ |
| Cone half-angle | $\theta$ | $20^{\circ}$ |

Treat the motion as non-relativistic for the kinematics: the point of the problem is the geometry of the three families of curves, not the Lorentz factor.

## Problem 3. Classifying the equations of compressible flow

A partial differential equation (PDE) has to be **classified** as the type fixes what data may be imposed, what the solution depends on, and whether the problem is to be solved by marching or by relaxation. In this problem you will classify the equation that governs small disturbances in a compressible flow, and watch it change type as the flow goes supersonic.

### The equation:

For a steady, 2D, irrotational flow of a perfect gas that differs only slightly from a uniform stream of Mach number $M_\infty$, the velocity potential $\phi$ of the disturbance obeys the **small-disturbance (Prandtl-Glauert) equation**:

$$\left(1 - M_\infty^{2}\right)\frac{\partial^{2}\phi}{\partial x^{2}} + \frac{\partial^{2}\phi}{\partial y^{2}} = 0$$

This is a particular case of the general second-order equation seen in the lectures,

$$A\,\phi_{xx} + B\,\phi_{xy} + C\,\phi_{yy} + D\,\phi_x + E\,\phi_y + F\phi + G = 0$$

whose type is decided by the discriminant $B^{2} - 4AC$, and whose characteristic directions satisfy:

$$A\left(\frac{dy}{dx}\right)^{2} - B\,\frac{dy}{dx} + C = 0$$
